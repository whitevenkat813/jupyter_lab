## **Data frame**

```python
import pandas as pd
```

```python
car_name = pd.Series(["TATA", "AUDI", "BMW", "TOYATO","HONDA"])
car_name
```

```python
car_color = pd.Series(["white","red","black","blue","Dgreen"])
car_color
```

```python
car_data = pd.DataFrame({
    "name" : car_name,
    "color": car_color
})
car_data
```

```python
print(car_data.dtypes)
print()
print(car_data.shape)
print()
print(car_data.size)
```

```python
car_data.info()
```

```python
car_data.isna()
```

```python
car_data.isna().sum()
```
