## **1D, 2D, 3D**

```python
array_1 = np.array([10,20,30,40])
print(array_1)
print (array_1.shape)
print (array_1.size)
```

```python
l1 = np.array([1,2,3,4,5])
l1 = np.append(l1,[6,7,8])
print(l1)
print(len(l1))
```

```python
l1 = [1,2,3,4,5]
l1.append([6,7,8])
print(l1)
print(len(l1))
```

```python
l1 = np.array([1,2,3,4,5])
l1 = np.insert(l1,2,[6,7,8])
print(l1)
print(len(l1))
```
