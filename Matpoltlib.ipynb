## **python Matplotlib**

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
plt.figure()
plt.plot()
plt.show();
```

## **scatter plot**

```python
x = np.array([1,2,3,4])
y = np.array([1,2,3,4])
c = ["purple","crimson","indigo","mediumblue"]

plt.scatter(x,y,marker="^",c= c,s=50);
```

## **line grap**

```python
x = np.array([1,2,3,4])
y = np.array([3,1,4,2])

plt.plot(x,y,linestyle="dashed")
plt.scatter(x,y,marker="p",c=["purple","crimson","gold","blue"])
plt.show();
```

```python
x = np.linspace(0, 10, 100)

plt.plot(x, np.sin(x), c="r", label="sin")
plt.plot(x, np.cos(x), c="b", label="cos")
plt.legend()
plt.show();
```

```python
import matplotlib.pyplot as plt

odometer = [150043, 87899, 32549, 11179, 213095, 99213, 45698, 54738, 60000, 31600]
price    = [4000, 5000, 7000, 22000, 3500, 4500, 7500, 7000, 6250, 9700]

plt.scatter(odometer, price, c="r", s=100)
plt.title("Price vs Odometer")
plt.xlabel("Odometer (KM)")
plt.ylabel("Price ($)")
plt.grid(True)
plt.show();
```

## **Bar chart**

```python
overs = ["1","2","3","4","5","6"]
runs = ["15","25","35","50","60","70"]

c=["rosybrown","lightcoral","r","indianred","brown","darkred"]

plt.bar(overs,runs,color=c,align="edge")
plt.title("PowerPlay Innings")
plt.xlabel("Innings overs")
plt.ylabel("Innings Total")

plt.show();
```

## **frquency distributtion**

```python
np.random.seed(2)
a = np.random.normal(50,2,100)

plt.hist(a,color="fuchsia",bins=15
         ,edgecolor="black")
plt.show();
```

## **pie chart**

```python
players = [
    "B Mccullum*",
        "Rihit",
           "virat",
        "Dhoni",         
    "H pandya"
]
runs = [152,
        100,
        90,
        30,
        25
]
exp = [
        .1,
        .1,
        0,
        0,
        0
]

plt.pie(runs,labels=players,explode=exp)
plt.title("Player of Match")
plt.show();
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, c="r")
plt.fill_between(x, y, color="red", alpha=0.3)
plt.show();
```

## **Legend**

```python
stackA = [1,2,3,4,6]
stackB = [3,4,5,6,7]
stackC = [6,5,4,3]


plt.plot(stackA,label="stockA",marker="o")
plt.plot(stackB,label="stockB",marker="*")
plt.plot(stackC,label="stockC",marker="P")

plt.legend()
plt.title("Stock Market")
plt.show();
```
